```python
import numpy as np
from datetime import datetime

# PARTICIPANTS

participants = [
    "Rahul",
    "Priya",
    "Neha",
    "Aman",
    "Karan",
    "Vikas"
]


# 6 x 24 MATRIX

activity = np.zeros((len(participants), 24), dtype=int)


# READ CHAT FILE

with open("hostel_bois.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()


# COUNT MESSAGES PER PERSON PER HOUR

for line in lines:

    line = line.strip()

    if " - " not in line:
        continue

    timestamp = line.split(" - ", 1)[0]
    remaining = line.split(" - ", 1)[1]

    if ": " not in remaining:
        continue

    person = remaining.split(": ", 1)[0]

    if person not in participants:
        continue

    date_time = datetime.strptime(
        timestamp,
        "%d/%m/%y, %H:%M"
    )

    hour = date_time.hour

    person_index = participants.index(person)

    activity[person_index][hour] += 1


# PRINT MATRIX

print("ACTIVITY MATRIX")
print()

for i in range(len(participants)):
    print(participants[i], activity[i])


# PRINT HEATMAP

print()
print("ACTIVITY HEATMAP")
print()

print("       ", end="")

for hour in range(24):
    print(f"{hour:2}", end=" ")

print()

maximum = activity.max()

for i in range(len(participants)):

    print(f"{participants[i]:6}", end=" ")

    for hour in range(24):

        value = activity[i][hour]

        if value == 0:
            symbol = " "

        elif value <= maximum * 0.25:
            symbol = "░"

        elif value <= maximum * 0.50:
            symbol = "▒"

        elif value <= maximum * 0.75:
            symbol = "▓"

        else:
            symbol = "█"

        print(f"{symbol:2}", end=" ")

    print()
```

```python
from datetime import datetime, timedelta


# READ FILE

with open("hostel_bois.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()


# PARSE MESSAGES

messages = []

for line in lines:

    line = line.strip()

    if " - " not in line:
        continue

    timestamp = line.split(" - ", 1)[0]
    remaining = line.split(" - ", 1)[1]

    if ": " not in remaining:
        continue

    person = remaining.split(": ", 1)[0]
    text = remaining.split(": ", 1)[1]

    messages.append([timestamp, person, text])


participants = [
    "Rahul",
    "Priya",
    "Neha",
    "Aman",
    "Karan",
    "Vikas"
]


# RESPONSE TIME

response = {}

for i in range(1, len(messages)):

    try:
        t1 = datetime.strptime(
            messages[i - 1][0],
            "%d/%m/%y, %H:%M"
        )

        t2 = datetime.strptime(
            messages[i][0],
            "%d/%m/%y, %H:%M"
        )

    except ValueError:
        continue

    person1 = messages[i - 1][1]
    person2 = messages[i][1]

    if person1 != person2:

        gap = (t2 - t1).total_seconds()

        if gap >= 0:

            if person2 not in response:
                response[person2] = []

            response[person2].append(gap)


print("RESPONSE PATTERNS")
print("--------------------")

for person in participants:

    if person in response:

        average = sum(response[person]) / len(response[person])

        if average >= 3600:
            print(
                person,
                ":",
                round(average / 3600, 1),
                "hours"
            )
        else:
            print(
                person,
                ":",
                round(average / 60, 1),
                "minutes"
            )


# ACTIVE DAYS

active = {}

for m in messages:

    try:
        date = datetime.strptime(
            m[0],
            "%d/%m/%y, %H:%M"
        ).date()

    except ValueError:
        continue

    person = m[1]

    if person not in participants:
        continue

    if person not in active:
        active[person] = set()

    active[person].add(date)


# START AND END DATE

all_dates = []

for person in active:
    for date in active[person]:
        all_dates.append(date)

start = min(all_dates)
end = max(all_dates)


# SILENT STREAK

print()
print("LONGEST SILENT STREAKS")
print("--------------------")

for person in participants:

    streak = 0
    longest = 0

    day = start

    while day <= end:

        if day not in active.get(person, set()):

            streak += 1

            if streak > longest:
                longest = streak

        else:

            streak = 0

        day += timedelta(days=1)

    print(person, ":", longest, "days")
```

```python
def spammer(data):
  return len(data)
```

```python
def group_mom(data):
  words = ['okay', 'safe', 'eat', 'sleep', 'take care', 'are you', 'please', 'reminder', 'drink water']

  count = 0

  for message in data:
    text = message[2].lower()

    for word in words:
      if word in text:
        count += 1

  return count
```

```python
def night_owl(data):
  count = 0

  for message in data:
    hour = int(message[0].split(",")[1].split(":")[0])
    if hour>=23 or hour<=24:
      count += 1
  return count
```

```python
def storyteller(data):
    total = 0

    for m in data:
        total += len(m[2].split())

    if len(data) > 0:
        return total / len(data)
    else:
        return 0
```

```python
def drama_queen(data):
    count = 0

    for m in data:
        text = m[2]

        if text.isupper() and any(char.isalpha() for char in text):
            count += 1

    return count
```

```python
def ghost(data):

    person = data[0][1]

    if person in active:
        return -len(active[person])

    return 0
```

```python
def comedian(data):
    count = 0

    for m in data:
        text = m[2].lower()

        if "haha" in text or "lol" in text or "lmao" in text:
            count += 1

    return count
```

```python
def question_master(data):

    count = 0

    for m in data:

        if "?" in m[2]:
            count += 1

    return count
```

```python
print("PERSONALITY ARCHETYPES")
print("----------------------")

for person in participants:

    data = person_data[person]

    scores = {
        "THE SPAMMER": spammer(data),
        "THE GROUP MOM": group_mom(data),
        "THE NIGHT OWL": night_owl(data),
        "THE STORYTELLER": storyteller(data),
        "THE DRAMA QUEEN": drama_queen(data),
        "THE GHOST": ghost(data),
        "THE COMEDIAN": comedian(data),
        "THE QUESTION MASTER": question_master(data)
    }

    result = max(scores, key=scores.get)

    print(person, "→", result)
```
